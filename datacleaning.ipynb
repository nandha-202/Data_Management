```python
import pandas as pd, numpy as np

data = pd.read_csv(r"C:\Users\Hp\Downloads\titanic (1)(in).csv")
data
```

```python
print(data.isnull().sum())#isnull - missing values 
#sum - count missing values in columns 
```

```python
# Age
data["age"] = data["age"].fillna(data["age"].median())

# Cabin
data["deck"] = data["deck"].fillna("Unknown")

# Embarked
data["embarked"] = data["embarked"].fillna(data["embarked"].mode())

print(data.isnull().sum())
print("total null vaues: ", (data.isnull().sum().sum()))
```

```python
duplication = data.dropna()
print(duplication)
```
