```python
import pandas as pd , numpy as np

data = pd.read_csv(r"C:\Users\Hp\Downloads\titanic (1)(in).csv")
data
```

```python
avg_age = data.groupby("survived")["age"].mean()

print(avg_age)
```

```python
print("Average age of non-survivors:", avg_age[0])
print("Average age of survivors:", avg_age[1])
```

```python
gender_survival = data.groupby("sex")["survived"].mean() * 100

print(gender_survival)
```

```python
class_survival = data.groupby("pclass")["survived"].mean() * 100

print(class_survival)
```

```python
embark_survival = data.groupby("embarked")["survived"].mean() * 100

print(embark_survival)
```
