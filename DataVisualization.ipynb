```python
import pandas as pd, matplotlib.pyplot as plt

data = pd.read_csv(r"C:\Users\Hp\Downloads\titanic (1)(in).csv")
print(data.columns)
data
```

```python
#bar chart
classs = data["pclass"].value_counts()
print(classs)
```

```python
plt.bar(classs.index,classs.values)
plt.title("distribution by class")
plt.xlabel("class")
plt.ylabel("no. of passengers")

plt.show()
```

```python
gender = data["sex"].value_counts()
print(gender)
```

```python
plt.bar(gender.index, gender.values, color="red")
plt.title("distribution by gender")
plt.xlabel("gender")
plt.ylabel("no. of passengers")
plt.savefig("plot.jpg")
plt.show()
```

```python
#histrogram
plt.hist(data['age'].dropna(), bins=10,color="pink")
plt.title("age distribution of titanic")
plt.xlabel("age")
plt.ylabel("no. of passenger")
plt.savefig("age_distributin.jpg", dpi =300)
plt.show()
```

```python
#scatter chart

scatter_data = data[['age', 'fare']].dropna()
plt.scatter(scatter_data['age'], scatter_data['fare'], color="yellow")
plt.title("age vs fare")
plt.xlabel("age")
plt.ylabel("fare")
plt.savefig("age_vs_fare.jpg")
plt.show
```
